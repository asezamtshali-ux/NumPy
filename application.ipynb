Loading the sheets using pandas

```python
tech_df = pd.read_excel(file_path, sheet_name='Technical_Assessment')
market_df = pd.read_excel(file_path, sheet_name='Market_Research')
```

data cleaning and preparation pipeline

```python
software_scores = pd.to_numeric(tech_df['Software_Score'], errors='coerce').to_numpy()
datascience_scores = pd.to_numeric(tech_df['DataScience_Score'], errors='coerce').to_numpy()
cyber_scores = pd.to_numeric(tech_df['Cybersecurity_Score'], errors='coerce').to_numpy()  
market_size = pd.to_numeric(market_df['Market_Size'], errors='coerce').to_numpy()
# Using pd.to_numeric to force text anomalies into NaN/blank values so NumPy can handle them
# errors = coerce error handling since i am still working with dirty data
```

```python
print(f"Average software_scores:      {np.nanmean(software_scores):.2f}")
print(f"Average cyber_scores: {np.nanmean(cyber_scores):.2f}")  
print(f"Median datascience_scores:   {np.nanmedian(datascience_scores):.2f}")
print(f"Top cyber_scores:     {np.nanmax(cyber_scores):.2f}")   
# nanmean ignoring the blank spaces
```

```python
top_tech_mask = (software_scores > 80) & (cyber_scores > 80) & (~np.isnan(software_scores)) & (~np.isnan(cyber_scores))
top_tech_count = np.sum(top_tech_mask)
print(f"Projects scoring > 80 in BOTH SW & Cyber: {top_tech_count}")
```

```python
print(f"Total Combined Market Size:  ${np.nansum(market_size):,.2f}")
print(f"Average Market Size:         ${np.nanmean(market_size):,.2f}")
```
